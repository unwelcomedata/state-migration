DuckDB query sandbox. Use this notebook to poke around loaded tables,
check schemas, preview data, and run ad-hoc queries without modifying
anything in the pipeline.

Nothing in this notebook affects the pipeline — it's read-only.

```python
import sys
from pathlib import Path
import pandas as pd
import duckdb

PROJECT = Path.cwd()
while not (PROJECT / "config.yaml").exists() and PROJECT != PROJECT.parent:
    PROJECT = PROJECT.parent

con = duckdb.connect(str(PROJECT / "data" / "project.duckdb"), read_only=True)

def q(sql):
    """Run a query and return a DataFrame."""
    return con.sql(sql).df()

print(f"Connected to: {PROJECT / 'data' / 'project.duckdb'}")
```

```python
# List all tables
q('SHOW TABLES')
```
